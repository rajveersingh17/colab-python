```python
import numpy as np

arr2 = np.array([[10, 20], [30, 40]])

print("Full 2D Array:")
print(arr2)

print("\nSpecific elements:")
print(arr2[1, 1])
print(arr2[0, 0])

print("\nSlicing:")
print(arr2[:, 1])

print("\nFancy Indexing:")
print(arr2[[0, 1], [1, 0]])
print(arr2[[0, 1], [0, 1]])

# Advanced operations on 2D
print("\nSum of all elements:", np.sum(arr2))
print("Mean of rows:", np.mean(arr2, axis=1))
print("Transpose:\n", arr2.T)
```

```python
import numpy as np

arr3 = np.array([[[10, 20], [30, 40]],
                 [[50, 60], [70, 80]]])

print("Full 3D Array:")
print(arr3)

print("\nAccessing element [0, 1, 1]:", arr3[0, 1, 1])

print("\nSlicing [:, :, 1] (Taking column 1 from all layers and rows):")
print(arr3[:, :, 1])

# Advanced 3D reshaping and manipulation
arr_reshaped = arr3.reshape(4, 2)
print("\nReshaped to 4x2:\n", arr_reshaped)

# Creating a larger 3D structure
large_3d = np.arange(27).reshape(3, 3, 3)
print("\nLarge 3x3x3 Array:")
print(large_3d)

# Masking/Filtering
mask = large_3d > 10
print("\nElements greater than 10:")
print(large_3d[mask])
```

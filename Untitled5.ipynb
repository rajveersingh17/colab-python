```python
class User1:
  def __init__(self,pen,pencil):
    self.pen = pen
    self.pencil = pencil
  def penGet(self):
    print("pen is working: ",self.pen)
  def pencilGet(self):
    print("pencil is working: ",self.pencil)

class User2(User1):
  def __init__(self,pen,pencil,earser):
    self.pen = pen
    self.pencil = pencil
    self.earser = earser
# object
u = User2("blue","black","doms")
u.penGet()
```

```python
class User1:
    def __init__(self, pen, pencil):
        print("constructor of user 1")
        self.pen = pen
        self.pencil = pencil
    def penGet(self):
        print("pen is working:", self.pen)
    def pencilGet(self):
        print("pencil is working:", self.pencil)
class User2(User1):
    def __init__(self, pen, pencil, earser):
        print("constructor of user 2")
        super().__init__(pen, pencil)
        self.earser = earser
# object
u = User2("blue", "black", "doms")
u.penGet()
```

```python
# encapsulation

class Student:
    def __init__(self, name, roll_no):
        self.name = name
        self.roll_no = roll_no
    def getname(self):
        print(self.name)
s = Student("Rahul", 123)
print(s.name)
print(s.roll_no)
```

```python
# Encapsulation
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    def showDetails(self):
        print("Employee Name:", self.name)
        print("Salary:", self.salary)
    def updateSalary(self, new_salary):
        self.salary = new_salary
        print("Salary Updated!")
# object
e = Employee("Rahul", 25000)
e.showDetails()
e.updateSalary(30000)
e.showDetails()
```

```python
# class variable
class Student:
    count = 0  # count is class variable
    def __init__(self, name, roll_no):
        print("constructor is working")
        self.name = name
        self.roll_no = roll_no
        Student.count += 12  # update class variable
    def getname(self):
        print(self.name)
s = Student("Rahul", 123)
s.getname()
print(s.count)
```

```python
# Method Overriding

class Suman:
    def __init__(self, pen, pencil):
        self.pen = pen
        self.pencil = pencil
    def getpen(self):
        print("Suman:", self.pen)
class Rahul(Suman):
    def __init__(self, pen, pencil, eraser):
        self.eraser = eraser
        super().__init__(pen, pencil)
    def getpen(self):
        print("Rahul:", self.pen)
u = Rahul("blue", "black", "doms")
u.getpen()
```

```python
class User1:
  def __init__(self,pen,pencil):
    self.pen = pen
    self.pencil = pencil
  def penGet(self):
    print("pen is working: ",self.pen)
  def pencilGet(self):
    print("pencil is working: ",self.pencil)

class User2(User1):
  def __init__(self,pen,pencil,earser):
    self.pen = pen
    self.pencil = pencil
    self.earser = earser
# object
u = User2("blue","black","doms")
u.penGet()
```

```python
class User1:
    def __init__(self, pen, pencil):
        print("constructor of user 1")
        self.pen = pen
        self.pencil = pencil
    def penGet(self):
        print("pen is working:", self.pen)
    def pencilGet(self):
        print("pencil is working:", self.pencil)
class User2(User1):
    def __init__(self, pen, pencil, earser):
        print("constructor of user 2")
        super().__init__(pen, pencil)
        self.earser = earser
# object
u = User2("blue", "black", "doms")
u.penGet()
```

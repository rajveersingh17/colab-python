```python
class User1:
  def __init__(self,pen,pencil):
    self.pen = pen
    self.pencil = pencil
  def penGet(self):
    print("pen is working: ",self.pen)
  def pencilGet(self):
    print("pencil is working: ",self.pencil)

class User2(User1):
  def __init__(self,pen,pencil,earser):
    self.pen = pen
    self.pencil = pencil
    self.earser = earser
# object
u = User2("blue","black","doms")
u.penGet()
```

```python
class User1:
    def __init__(self, pen, pencil):
        print("constructor of user 1")
        self.pen = pen
        self.pencil = pencil
    def penGet(self):
        print("pen is working:", self.pen)
    def pencilGet(self):
        print("pencil is working:", self.pencil)
class User2(User1):
    def __init__(self, pen, pencil, earser):
        print("constructor of user 2")
        super().__init__(pen, pencil)
        self.earser = earser
# object
u = User2("blue", "black", "doms")
u.penGet()
```

```python
class Suman:
    def __init__(self, bike):
        self.bike = bike
    def usebike(self):
        print("bike used by:", self.bike)
class Rahul(Suman):
    def __init__(self, name, bike):
        self.name = name  # name is attribute
        # approach one
        # self.bike = bike
        # approach second
        super().__init__(bike)
    def usename(self):
        print("name is:", self.name)
u = Rahul("Rahul Sharma", "TVS Raider")
u.usename()
u.usebike()
```

```python
# encapsulation

class Student:
    def __init__(self, name, roll_no):
        self.name = name
        self.roll_no = roll_no
    def getname(self):
        print(self.name)
s = Student("Rahul", 123)
print(s.name)
print(s.roll_no)
```

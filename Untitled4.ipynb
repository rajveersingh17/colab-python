```python
def students(*data):
    print(data)

students('Rahul', 'Aman')
students('Python', 'Java', 'C++', 'HTML', 'CSS')
students('Hello')
```

```python
def fruits(*items):
    for i in items:
        print(i)

fruits('Apple', 'Mango')
fruits('Banana', 'Orange', 'Grapes', 'Papaya')
fruits('Kiwi')
```

```python
def numbers(*num):
    total = 0

    for i in num:
        total = total + i

    print("Sum =", total)

numbers(10, 20)
numbers(5, 10, 15, 20)
numbers(100)
```

```python
def atm(a, b):
    print(a + b)

pocket = atm(200, 500)
print(pocket)
```

```python
import numpy as np

arr = np.array([10, 12, 15, 18, 20, 19, 30, 11, 17, 14])

arr_f = arr[(arr > 10) & (arr < 20)]

print(arr_f)
```

```python
# where
import numpy as np

arr = np.array([10, 20, 3, 4, 90, 100])
print(arr)

arr_w = np.where(arr > 9, "True", "False")
print(arr_w)

# question: numbers from 10 to 20 -> multiply by 2, otherwise add 2

arr1 = np.array([5, 6, 7, 8, 11, 24, 16, 19, 25])

result = np.where((arr1 >= 10) & (arr1 <= 20), arr1 * 2, arr1 + 2)

print(arr1)
print(result)
```

```python
#Sum of all elements
import numpy as np

a = np.array([10, 20, 30, 40, 50])

print(np.sum(a))
```

```python
#Average of all elements
import numpy as np

a = np.array([10, 20, 30, 40, 50])

print(np.mean(a))
```

```python
import numpy as np

a = np.array([10, 20, 30, 40, 50])

print("Array =", a)
print("Sum =", np.sum(a))
print("Mean =", np.mean(a))
print("Median =", np.median(a))
print("Minimum =", np.min(a))
print("Maximum =", np.max(a))
```

```python
import numpy as np

a = np.array([10, 20, 30, 40, 50, 20, -10])

print("Array =", a)

print("Variance =", np.var(a))
print("Standard Deviation =", np.std(a))
print("Product =", np.prod(a))
print("Cumulative Sum =", np.cumsum(a))
print("Cumulative Product =", np.cumprod(a))
print("Index of Minimum Value =", np.argmin(a))
print("Index of Maximum Value =", np.argmax(a))
print("Absolute Values =", np.abs(a))
print("Unique Values =", np.unique(a))
print("50th Percentile =", np.percentile(a, 50))
print("0.5 Quantile =", np.quantile(a, 0.5))
print("Range (Max - Min) =", np.ptp(a))
print("Any Non-Zero Value =", np.any(a))
```

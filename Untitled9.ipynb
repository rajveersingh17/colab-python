```python
import numpy as np

arr = np.array([10, 12, 15, 18, 20, 19, 30, 11, 17, 14])

arr_f = arr[(arr > 10) & (arr < 20)]

print(arr_f)
```

```python
# where
import numpy as np

arr = np.array([10, 20, 3, 4, 90, 100])
print(arr)

arr_w = np.where(arr > 9, "True", "False")
print(arr_w)

# question: numbers from 10 to 20 -> multiply by 2, otherwise add 2

arr1 = np.array([5, 6, 7, 8, 11, 24, 16, 19, 25])

result = np.where((arr1 >= 10) & (arr1 <= 20), arr1 * 2, arr1 + 2)

print(arr1)
print(result)
```

```python
import numpy as np

arr1 = np.array([[10, 20],
                 [30, 40]])

arr2 = np.array([[1, 2],
                 [3, 4]])

arr3 = np.concatenate((arr1, arr2), axis=0)

print(arr3)
```

```python
#Sum of all elements
import numpy as np

a = np.array([10, 20, 30, 40, 50])

print(np.sum(a))
```

```python
#Average of all elements
import numpy as np

a = np.array([10, 20, 30, 40, 50])

print(np.mean(a))
```

```python
#Middle value after sorting
import numpy as np

a = np.array([50, 10, 40, 20, 30])

print(np.median(a))
```

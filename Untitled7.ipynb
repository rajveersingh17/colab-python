```python
# numpy | list
import numpy as np
import time

# for list
start = time.time()

sum1 = 0
for i in range(100000):
    sum1 += i * 2

end = time.time()
print("for list", sum1, end - start)

# for numpy | array
start1 = time.time()

arr = np.arange(100000) * 2
sum2 = arr.sum()

end1 = time.time()
print("for array", sum2, end1 - start1)
```

```python
# zeros
import numpy as np

# array 1d
arr = np.zeros(5)
print(arr)

# 2d
arr1 = np.zeros((2, 3))   # 2 rows and 3 columns
print(arr1)
```

```python
# ones
import numpy as np

# array 1d
arr = np.ones(5)
print(arr)

# 2d
arr1 = np.ones((2, 3))   # 2 rows and 3 columns
print(arr1)
```

```python
# full
import numpy as np

# array 1d
arr = np.full((5), 7)
print(arr)

# 2d
arr1 = np.full((2, 3), 7)
print(arr1)
```

```python
# array opr on zeros | ones | full
import numpy as np

arr = np.zeros(5)
print(arr)

print(arr.shape)   # (5,)
print(arr.ndim)    # 1d
print(arr.size)    # 5
print(arr.dtype)   # float64

# 2d
print("2d ones start")

arr1 = np.zeros((3, 2))
print(arr1)

print(arr1.shape)  # (3,2)
print(arr1.ndim)   # 2d
print(arr1.size)   # 6
print(arr1.dtype)  # float64
```

```python
# 1d array
import numpy as np

arr = np.arange(5)
print(arr)

# 2d array
arr1 = np.arange(6).reshape(2, 3)
print(arr1)
```

```python
import numpy as np

#  2D array
arr = np.arange(6).reshape(2, 3)
print(arr)

print(arr.ravel())
```

```python
import numpy as np

arr = np.arange(6).reshape(2, 3)
print(arr)

arr_new1 = arr.ravel()
arr_new1[0] = 200
print(arr)
print(arr_new1)
```
